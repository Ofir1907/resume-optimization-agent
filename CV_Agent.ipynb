```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
system_instruction = """
You are an expert technical career strategist and resume optimization specialist.

Your task is to analyze a candidate's comprehensive master background and tailor it into a strictly focused, high-impact 1-page resume matching a target job description.

Follow these strict operating rules:
1. Ground Truth & Strict Role Integrity:
   - Use ONLY experiences, achievements, credentials, and technical skills present in the base resume. Never invent tools, metrics, or positions.
   - Never attribute a skill, tool, or achievement to a role where it was not explicitly performed in the base resume.
   - Even if a candidate has a skill listed elsewhere, you may ONLY mention it within the specific roles or projects where it was explicitly practiced.
2. Aggressive Relevance Filtering (1-Page Fit):
   - You MUST select EXACTLY 3 distinct work experiences from the base resume that best align with the target job requirements.
   - Rank and pick roles prioritizing quantitative/statistical analysis, data pipelines, and software development.
   - For each of the 3 selected roles, provide exactly 2 to 3 concise, high-impact bullet points.
3. Keyword Alignment Without Fabrication:
   - Rephrase bullets solely to highlight impact and relevance using active verbs and terminology present in the original experience.
   - Do NOT inject missing technologies or stack items from the job description into past roles to force a keyword match.
4. Skills Curation: Under professional_skills, list only tools and skills directly relevant to the role (omit irrelevant tools to save vertical space).
"""
```

```python
cv_prompt_template = ChatPromptTemplate.from_messages([
    ("system", system_instruction.strip()),
    ("human", """
Candidate Full Background & Base Resume:
\"\"\"
{base_cv}
\"\"\"

Target Job Description:
\"\"\"
{job_description}
\"\"\"

Please optimize, prioritize, and generate the tailored 1-page resume strictly according to the defined JSON schema.
""")
])
```

```python
from langchain_groq import ChatGroq

# 1. define the LLM with the desired model and temperature
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.1,
    max_tokens=4096,
)

# 2. link the prompt template you created in step 1 directly to the model
cv_chain = cv_prompt_template | llm
```

```python
from pydantic import BaseModel, Field
from typing import List, Optional

# Define the Pydantic models to match the expected JSON schema;
class ContactInfo(BaseModel):
    phone: str
    linkedin: str
    email: str
    github: str

class PersonalInfo(BaseModel):
    name: str
    title: str
    contact: ContactInfo

class WorkExperienceItem(BaseModel):
    role: str
    organization: str
    period: str
    highlights: List[str]

class TechnicalProjectItem(BaseModel):
    name: str
    technologies: str
    highlights: List[str]

class EducationItem(BaseModel):
    degree: str
    institution: str
    period: str
    honors_or_notes: Optional[str] = None

class MilitaryService(BaseModel):
    role: str
    period: str
    description: str

class LanguageItem(BaseModel):
    language: str
    proficiency: str

class ProfessionalSkills(BaseModel):
    languages: List[str] = Field(default_factory=list)
    software_development: List[str] = Field(default_factory=list)
    data_analysis_and_visualization: List[str] = Field(default_factory=list)
    environments: Optional[List[str]] = Field(default_factory=list)
    bioinformatics_and_domain_tools: Optional[List[str]] = Field(default_factory=list)
    machine_learning: Optional[List[str]] = Field(default_factory=list)
    soft_skills: Optional[List[str]] = Field(default_factory=list)

class TailoredResume(BaseModel):
    personal_info: PersonalInfo
    about_me: str
    work_experience: List[WorkExperienceItem]
    technical_projects: Optional[List[TechnicalProjectItem]] = []
    professional_skills: ProfessionalSkills
    education: List[EducationItem]
    military_service: Optional[MilitaryService] = None
    languages: List[LanguageItem]

# adding structured output to the LLM to ensure it returns a validated Pydantic object
structured_llm = llm.with_structured_output(TailoredResume, method="json_mode")

# create a new chain that combines the prompt template with the structured LLM output
cv_chain = cv_prompt_template | structured_llm

result = cv_chain.invoke({
    "base_cv": my_base_cv,
    "job_description": target_job_description
})

```
